```python
import pandas as pd
import numpy as np
from pathlib import Path
from scripts import *

p = Path('genai_results_new/vlm')
dfs = {}
for f in p.iterdir():
    if 'batched' or 'batch' in str(f):
        dfs[str(f)] = pd.read_csv(f, nrows=9)
```

# INT4

```python
cpu = prepare_benchmark(dfs['genai_results_new/vlm/vlm_cpu_int4_batched.csv'])
gpu = prepare_benchmark(dfs['genai_results_new/vlm/vlm_gpu_int4_batched.csv'])

int4 = {
    "cpu": cpu,
    "gpu": gpu,
}

data_int4 = pd.concat(
    [prepare_benchmark(df).assign(source=name)
     for name, df in int4.items()],
    ignore_index=True,
)
```

```python
plot_generation_throughput(data_int4)
```

```python
plot_startup_cost(int4)
```

```python
plot_latency(data_int4)
```

```python
plot_ttft(int4)
```

```python
plot_tpot(int4)
```
