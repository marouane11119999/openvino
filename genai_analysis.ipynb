```python
import pandas as pd
import numpy as np
from pathlib import Path
from scripts import *

p = Path('genai_results/text_gen')
dfs = {}
for f in p.iterdir():
    dfs[str(f)] = pd.read_csv(f, nrows=9)
```

# Benchmarking LLMS : TEXT GENERATION

```python
qwen_7b_cpu = prepare_benchmark(dfs['genai_results/text_gen/qwen2.5-7B-Instruct-int4-ov_CPU.csv'])
qwen_7b_gpu = prepare_benchmark(dfs['genai_results/text_gen/qwen2.5-7B-Instruct-int4-ov_GPU.csv'])
qwen_7b_npu = prepare_benchmark(dfs['genai_results/text_gen/qwen2.5-7B-Instruct-int4-ov_NPU.csv'])
qwen_7b = {
    "cpu": qwen_7b_cpu,
    "gpu": qwen_7b_gpu,
    "npu": qwen_7b_npu,
}

data_7b = pd.concat(
    [prepare_benchmark(df).assign(source=name)
     for name, df in qwen_7b.items()],
    ignore_index=True,
)
```

```python
plot_generation_throughput(data_7b)
```
